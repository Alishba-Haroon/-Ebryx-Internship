## _______________________ Pandas Series____________________________

### What is Pandas

Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool,
built on top of the Python programming language.

### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

# Importing Pandas

```python
import pandas as pd
import numpy as np
```

### Series from lists

```python
# string
country = ["Pakistan", "UK", "USA", "south Korea", "Turkey"]
pd_series = pd.Series(country)
pd_series
```

```python
# Integer
runs = [10,20,30,40,50]
run = pd.Series(runs)
run
```

```python
# custom index
marks = [50, 80, 66, 79, 34]
subject = ["urdu", "english", "math", "computer Science", "Chemistry"]

custom_index = pd.Series(marks, index=subject)
custom_index
```

```python
# setting a name
```

```python
custom_index = pd.Series(marks, index=subject, name = 'Ali Marks')
custom_index
```

### Series from dict

**Why does negative indexing not work in a Series and DataFrames?**

Pandas Series and DataFrames do not support direct negative indexing via the standard brackets [] because index labels can themselves be negative integers, creating ambiguity between a position from the end and an explicit label name


```python
# negative indexing 
# x[-1] give error

```

```python
temp = pd.Series([1,2,3,np.nan,5,6,np.nan,8,np.nan,10])
temp
```

```python
temp.size # tell all value size include null values
temp.count # give you value without null values
```

```python
# tell null values
temp.isnull().sum()
```

```python
# dropna
temp.dropna()
```

```python
# fillna
temp.fillna(temp.mean())
```
