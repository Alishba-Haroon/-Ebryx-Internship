# Linear Regression Using iris dataset

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
species_map = {
    'setosa': 0,
    'versicolor': 1,
    'virginica': 2
}

df['species_num'] = df['species'].map(species_map)

plt.scatter(df['petal_width'], df['species_num'])

plt.xlabel('Petal Width')
plt.ylabel('Species')
plt.yticks([0, 1, 2], ['setosa', 'versicolor', 'virginica'])

plt.show()
```

```python
X = df.iloc[:,0:1]
y = df.iloc[:,-1]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
X_test
```

```python
y_test
```

```python
lr.predict(X_test.iloc[0].values.reshape(1,1))
```

```python
plt.scatter(df['petal_width'], df['species_num'])

plt.plot(
    X_train,
    lr.predict(X_train),
    color='red'
)

plt.xlabel('Petal Width')
plt.ylabel('Species')

plt.yticks([0, 1, 2], ['setosa', 'versicolor', 'virginica'])

plt.show()
```

```python
m = lr.coef_
```
