```python
import seaborn as sns
import matplotlib as plt
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
```

```python
# scatter plot -> axes level function -> rectangle
sns.scatterplot(data=tips, x = 'total_bill', y = 'tip')
```

```python
# relplot is figure level -> square shape -> lib say use figure level function
sns.relplot(data=tips, x = 'total_bill', y = 'tip', kind='scatter', hue='sex', style='time', size='size')
```

```python
gap = px.data.gapminder()
temp_df = gap[gap['country'] == 'Nepal']
temp_df
```

```python
sns.lineplot(data=temp_df, x='year',y='lifeExp')
```

```python
sns.relplot(data=temp_df, x='year',y='lifeExp', kind='line')
```

```python
gap = px.data.gapminder()
```

```python
gap1 = gap[gap['country'].isin(['Nepal', 'Germany','China'])]
gap1
```

```python
sns.relplot(data=gap1, x='year',y='lifeExp', kind='line', hue='country', style='continent')
```

```python
# facet plot -> figure level function -> work with relplot
# it will not work with scatterplot and lineplot
sns.relplot(data=tips, x='total_bill', y='tip', col='sex', row='day')
```

```python
sns.relplot(data=tips, x='total_bill', y='tip', col='sex', row='day',kind='line')
```

```python
# col wrap
sns.relplot(data=gap, x='lifeExp', y='gdpPercap', kind='scatter', col='year', col_wrap=3)
```
